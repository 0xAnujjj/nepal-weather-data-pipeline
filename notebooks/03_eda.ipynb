### imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/clean/kat_clean.csv")
df.head()
```

### convert the data per day to months:

```python
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

profile = df.groupby("month").agg(
    temp_max_c=("temp_max_c", "mean"),
    temp_min_c=("temp_min_c", "mean"),
    temp_mean_c=("temp_mean_c", "mean"),
    wind_max_kmh=("wind_max_kmh", "mean"),
)
```

### time-series line plots

```python
df.set_index("date").plot(figsize=(12,6))
plt.title("Daily Weather Trends")
plt.ylabel("Value")
plt.show()
```

### temperature analysis (line chart: trend over time)

```python
plt.figure(figsize=(10, 4))
plt.plot(df["date"], df["temp_mean_c"])
plt.title("Daily Mean Temperature, Kathmandu")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.show()
```

### Wind analysis

```python
sns.lineplot(data=df, x="date", y="wind_max_kmh")
plt.title("Daily Maximum Wind Speed")
plt.xlabel("Date")
plt.ylabel("Wind Speed (km/h)")
plt.show()
```
